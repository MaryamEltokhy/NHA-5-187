```python
import os, subprocess

def show_disk():
    print(subprocess.run(["df", "-h", "/content"], capture_output=True, text=True).stdout)

def download_kaggle(dataset_slug, dest_path):

    os.makedirs(dest_path, exist_ok=True)
    subprocess.run(["kaggle", "datasets", "download", "-d", dataset_slug, "-p", dest_path], check=True)

def extract_and_cleanup(zip_path, extract_dir, delete_zip=True):
    if not os.path.exists(zip_path):
        print(f"[!] Not found: {zip_path}")
        return
    os.makedirs(extract_dir, exist_ok=True)
    print(f"Extracting {os.path.basename(zip_path)} ...")
    subprocess.run(["unzip", "-q", zip_path, "-d", extract_dir], check=True)
    count = len(os.listdir(extract_dir))
    print(f"  -> {count} items in {extract_dir}")
    if delete_zip:
        os.remove(zip_path)
        print(f"  -> deleted {os.path.basename(zip_path)} to free space")

def download_direct(url, dest_zip):
    os.makedirs(os.path.dirname(dest_zip), exist_ok=True)
    subprocess.run(["wget", "-q", "--show-progress", "-O", dest_zip, url], check=True)
```
